```python
import pandas as pd
import numpy as np
```

```python
# Read clean churn file
df = pd.read_csv('churn_clean.csv')
```

```python
df.info()
```

```python
df.duplicated() 
```

```python
df.isnull().sum()
```

```python
# Replace missing values in 'InternetService' with None
df['InternetService'] = df['InternetService'].fillna('None')
```

```python
df.isnull().sum()
```

```python
# Drop irrelevant columns
df = df.drop(columns=['CaseOrder', 'Interaction', 'UID', 'Customer_id', 'TimeZone', 'Job','Email', 
                      'Contacts', 'Item1', 'Item2', 'Item3', 'Item4', 'Item5', 'Item6', 'Item7', 'Item8'])
```

```python
df.info()
```

```python
# Convert 'Tenure' to int
df['Tenure'] = df['Tenure'].round().astype(int)
```

```python
# Round 'MonthlyCharge'
df['MonthlyCharge'] = df['MonthlyCharge'].round(2)
```

```python
print(df)
```

```python
# Save cleaned dataset
df.to_csv('d210_churn_cleaned.csv')
```

```python
census = pd.read_csv('acs2017_census_tract_data.csv')
```

```python
census.info()
```

```python
census = census.drop(columns=['TractId', 'Men', 'Women', 'VotingAgeCitizen', 'IncomeErr', 'IncomePerCap', 'IncomePerCapErr'])
```

```python
census.isnull().sum()
```

```python
census.duplicated()
```

```python
# Replace all other nulls with 0
census = census.fillna(0)

# Then fix Income using median
census['Income'] = census['Income'].replace(0, census['Income'].median())
```

```python
print(census)
```

```python
# Remove the word "County" (and optional extra spaces) from the County column
census['County'] = census['County'].str.replace(' County', '', regex=False).str.strip()

# Remove the word "Municipio" (and optional extra spaces) from the County column
census['County'] = census['County'].str.replace(' Municipio', '', regex=False).str.strip()
```

```python
census['County'].head()
```

```python
# Create a mapping of full state names to abbreviations
state_abbr = {
    'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR',
    'California': 'CA', 'Colorado': 'CO', 'Connecticut': 'CT', 'Delaware': 'DE',
    'Florida': 'FL', 'Georgia': 'GA', 'Hawaii': 'HI', 'Idaho': 'ID',
    'Illinois': 'IL', 'Indiana': 'IN', 'Iowa': 'IA', 'Kansas': 'KS',
    'Kentucky': 'KY', 'Louisiana': 'LA', 'Maine': 'ME', 'Maryland': 'MD',
    'Massachusetts': 'MA', 'Michigan': 'MI', 'Minnesota': 'MN', 'Mississippi': 'MS',
    'Missouri': 'MO', 'Montana': 'MT', 'Nebraska': 'NE', 'Nevada': 'NV',
    'New Hampshire': 'NH', 'New Jersey': 'NJ', 'New Mexico': 'NM', 'New York': 'NY',
    'North Carolina': 'NC', 'North Dakota': 'ND', 'Ohio': 'OH', 'Oklahoma': 'OK',
    'Oregon': 'OR', 'Pennsylvania': 'PA', 'Puerto Rico': 'PR', 'Rhode Island': 'RI',
    'South Carolina': 'SC', 'South Dakota': 'SD', 'Tennessee': 'TN',
    'Texas': 'TX', 'Utah': 'UT', 'Vermont': 'VT', 'Virginia': 'VA',
    'Washington': 'WA', 'West Virginia': 'WV', 'Wisconsin': 'WI', 'Wyoming': 'WY'
}

# Replace state with abbreviations
census['State'] = census['State'].map(state_abbr)
```
